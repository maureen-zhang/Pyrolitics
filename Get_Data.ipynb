```python
import os
import gdown

FOLDER_ID = "1P3cR7Ws4VRPZ3v8o1oLa-rk8K-Kxq-ar"
TARGET_DIR = "data/raw"

# Make sure the target directory exists
os.makedirs(TARGET_DIR, exist_ok=True)

# Construct the full Drive folder URL
url = f"https://drive.google.com/drive/folders/{FOLDER_ID}"

# Download all files from the folder
gdown.download_folder(
    url=url,
    output=TARGET_DIR,
    quiet=False,    # Set to True if you want less output
    use_cookies=False
)

print(f"✅ Data downloaded to {TARGET_DIR}")
```
